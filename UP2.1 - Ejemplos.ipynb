## 1. Arrays

```python
import numpy as np
import numpy.typing as npt
```

```python
notas_lista: list[float] = [7.5, 4.0, 9.0, 6.5, 3.0]
print([n * 10 for n in notas_lista])
```

```python
notas: npt.NDArray[np.float64] = np.array([7.5, 4.0, 9.0, 6.5, 3.0])
print(notas * 10)
print(notas.mean())
```

Lista frente a array con un millón de valores. Cada operación se repite cinco veces y se toma el mejor tiempo; aun así, cambia en cada ejecución y en cada ordenador.

```python
import timeit

lista: list[float] = [float(i) for i in range(1_000_000)]
array: npt.NDArray[np.float64] = np.array(lista)

# Cada operación se repite 5 veces y se queda el mejor tiempo
t_lista: float = min(timeit.repeat(lambda: [x * 1.21 for x in lista], number=1, repeat=5))
t_array: float = min(timeit.repeat(lambda: array * 1.21, number=1, repeat=5))

print(f"Lista: {t_lista:.4f} s · Array: {t_array:.4f} s")
print(f"NumPy es {t_lista / t_array:.0f} veces más rápido")
```

**Curiosidad:** cuánta memoria ocupa un millón de enteros en una lista y en un array. Lo de la lista es aproximado: Python comparte los enteros pequeños entre variables.

```python
import sys
enteros: list[int] = list(range(1_000_000))
numeros: npt.NDArray[np.int64] = np.arange(1_000_000)
bytes_lista: int = sys.getsizeof(enteros)
bytes_lista += sum(sys.getsizeof(n) for n in enteros)
print(f"Lista: {bytes_lista / 1e6:.0f} MB")
print(f"Array: {numeros.nbytes / 1e6:.0f} MB")
```

### Referencia: type hints y arrays generados

| Type hint | Valores |
|---|---|
| `npt.NDArray[np.float64]` | Decimales |
| `npt.NDArray[np.int64]` | Enteros |
| `npt.NDArray[np.bool_]` | True o False |
| `npt.NDArray[np.str_]` | Texto |

| Función | Qué genera |
|---|---|
| `np.zeros(n)` · `np.ones(n)` | n ceros o n unos |
| `np.arange(inicio, fin, paso)` | Como `range()`: el fin no entra |
| `np.linspace(inicio, fin, n)` | n valores repartidos por igual; el fin sí entra |

## 2. Operar sin bucles

```python
print(notas + 0.5)
print(notas * 10)
print(notas ** 2)
```

```python
teoria: npt.NDArray[np.float64] = np.array([7.0, 4.0, 9.0, 6.0])
practica: npt.NDArray[np.float64] = np.array([8.0, 6.0, 10.0, 3.0])
print(teoria + practica)
print(teoria * 0.4 + practica * 0.6)
```

### Referencia: funciones y agregaciones

| Código | Qué hace |
|---|---|
| `np.round(a, n)` | Redondea cada valor a n decimales |
| `np.sqrt(a)` · `np.abs(a)` | Raíz cuadrada · valor absoluto |
| `np.clip(a, mínimo, máximo)` | Deja cada valor entre un mínimo y un máximo |
| `np.exp(a)` · `np.log(a)` | Exponencial · logaritmo |
| `a.sum()` · `a.mean()` · `np.median(a)` | Suma · media · mediana |
| `a.min()` · `a.max()` | Mínimo · máximo |
| `a.argmin()` · `a.argmax()` | Posición del mínimo · del máximo |
| `a.std()` | Desviación típica (se explica en UP2.4) |
| `np.sort(a)` | Array ordenado; el original no cambia |

## 3. Seleccionar y filtrar

```python
print(notas[0], notas[-1])
print(notas[1:4])
print(notas[[0, 3]])
```

```python
aprobados: npt.NDArray[np.bool_] = notas >= 5
print(aprobados)
print(aprobados.sum(), aprobados.mean())
```

```python
nombres: npt.NDArray[np.str_] = np.array(
    ["Ada", "Linus", "Grace", "Edgar", "Alan"]
)
print(notas[aprobados])
print(nombres[notas >= 5])
```

```python
print(notas[(notas >= 5) & (notas < 9)])
print(notas[(notas < 4) | (notas >= 9)])
print(notas[~aprobados])
```

```python
estado: npt.NDArray[np.str_] = np.where(notas >= 5, "apto", "no apto")
print(estado)
recuperadas: npt.NDArray[np.float64] = notas.copy()
recuperadas[recuperadas < 5] = 5.0
print(recuperadas)
```

```python
edades: npt.NDArray[np.int64] = np.array([36, 28, 45, 52])
trozo: npt.NDArray[np.int64] = edades[:2]
trozo[0] = 99
print(edades)
```

```python
copia: npt.NDArray[np.int64] = edades[:2].copy()
copia[0] = 0
print(edades)
```

### Referencia: seleccionar y filtrar

| Código | Qué hace |
|---|---|
| `a[i]` · `a[inicio:fin]` · `a[[i, j]]` | Una posición · un tramo · varias posiciones |
| `a >= 5` | Array de True y False |
| `a[a >= 5]` | Máscara: solo los valores que cumplen |
| `(a >= 5) & (a < 9)` · `\|` · `~` | Y · o · no, cada condición entre paréntesis |
| `np.where(condición, x, y)` | Array nuevo: x donde se cumple, y donde no |
| `a[máscara] = valor` | Cambia esos valores en el propio array |
| `a[:2].copy()` | Copia independiente de un tramo (un tramo sin copiar es una vista) |

## 4. Tablas y datos que faltan

```python
examenes: npt.NDArray[np.float64] = np.array([
    [7.5, 8.0, 6.5],
    [4.0, 5.5, 3.0],
    [9.0, 9.5, 10.0],
    [6.5, 4.0, 7.0],
])
print(examenes.shape)
```

```python
print(examenes[0])
print(examenes[0, 1])
print(examenes[:, 0])
print(examenes[1:3, :2])
```

```python
print(examenes.mean().round(2))
print(examenes.mean(axis=0).round(2))
print(examenes.mean(axis=1).round(2))
```

```python
con_hueco: npt.NDArray[np.float64] = np.array([7.5, np.nan, 9.0, 6.5])
print(con_hueco.mean(), np.nan == np.nan)

faltan: npt.NDArray[np.bool_] = np.isnan(con_hueco)
print(faltan, faltan.sum())
print(np.nanmean(con_hueco).round(2))
print(np.where(faltan, 0.0, con_hueco))
```

```python
edades_ok: npt.NDArray[np.int64] = np.array([36, 28, 45])
edades_hueco: npt.NDArray[np.float64] = np.array([36, np.nan, 45])
print(edades_ok, edades_ok.dtype)
print(edades_hueco, edades_hueco.dtype)
```

### Referencia: tablas y nan

| Código | Qué hace |
|---|---|
| `t.shape` | (filas, columnas) |
| `t[fila, columna]` · `t[:, 0]` · `t[1:3, :2]` | Una celda · una columna · un trozo |
| `t.mean(axis=0)` | Un resultado por columna |
| `t.mean(axis=1)` | Un resultado por fila |
| `np.isnan(a)` | True donde falta el dato |
| `np.nanmean(a)` · `np.nansum(a)` · `np.nanmax(a)` | Calcular sin contar los huecos |
| `np.where(np.isnan(a), 0.0, a)` | Rellenar los huecos |
