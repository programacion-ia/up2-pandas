Los datos son **simulados**: los barrios de Benidorm son reales, pero los anuncios y los precios se han generado con NumPy.

```python
from pathlib import Path

import pandas as pd
```

## 1. Series

```python
precios: pd.Series = pd.Series(
    [950, 1200, 780],
    index=["Levante", "Poniente", "Foietes"],
    name="precio",
)
print(precios)
```

```python
print(precios["Poniente"])
print(precios.index)
```

```python
print(precios * 12)
print(precios[precios > 900])
```

```python
febrero: pd.Series = pd.Series(
    [1250, 800, 900],
    index=["Poniente", "Foietes", "Centro"],
)
print(precios + febrero)
```

## 2. DataFrame

```python
pisos_mini: pd.DataFrame = pd.DataFrame({
    "barrio": ["Levante", "Poniente", "Foietes", "Centro"],
    "metros": [70, 95, 60, 82],
    "precio": [950, 1200, 780, 1010],
})
print(pisos_mini)
```

```python
print(type(pisos_mini["precio"]))
print(pisos_mini[["barrio", "precio"]])
```

```python
print(pisos_mini.shape)
print(pisos_mini.columns)
print(pisos_mini.dtypes)
```

```python
print(pd.Series(["Levante", "Centro"]).dtype)
print(pd.Series([950, None]).dtype)
print(pd.Series(["Levante", 950]).dtype)
```

```python
def precio_medio(tabla: pd.DataFrame) -> float:
    return float(tabla["precio"].mean())

print(precio_medio(pisos_mini))
```

### Referencia: tipos de dato

| dtype | Contiene |
|---|---|
| `int64` | Enteros |
| `float64` | Decimales, o enteros con huecos |
| `bool` | True o False |
| `str` | Texto (en tutoriales anteriores a pandas 3 aparece como `object`) |
| `object` | Una mezcla de tipos |

## 3. Leer un CSV y primer vistazo

```python
ruta: Path = Path("datos") / "pisos.csv"
pisos: pd.DataFrame = pd.read_csv(ruta)
print(pisos.shape)
print(pisos.head(3))
```

```python
columnas: list[str] = ["barrio", "tipo", "metros", "precio"]
print(pisos[columnas].sample(3, random_state=1))
```

```python
pisos.info()
```

```python
print(pisos[["metros", "precio"]].describe().round(1))
```

```python
print(pisos.describe(include=["str"]))
```

```python
print(pisos["tipo"].value_counts())
```

```python
print(pisos["barrio"].nunique())
print(pisos["anunciante"].unique())
```

```python
huecos: pd.Series = pisos.isna().sum()
print(huecos[huecos > 0])
print(round(pisos["planta"].isna().mean(), 3))
```

### Referencia: leer y explorar un archivo

| Código | Para qué sirve |
|---|---|
| `pd.read_csv(ruta)` | Leer un CSV en un DataFrame |
| `pisos.shape` · `pisos.columns` · `pisos.dtypes` | Tamaño, nombres y tipos de las columnas |
| `pisos.head(n)` · `pisos.tail(n)` | Las n primeras o últimas filas (5 si no se indica n) |
| `pisos.sample(n, random_state=1)` | n filas al azar; `random_state` es la semilla |
| `pisos.info()` | Tipos y valores no vacíos de todas las columnas |
| `pisos.describe()` | count, mean, std, min, 25 %, 50 % (mediana), 75 %, max |
| `pisos.describe(include=["str"])` | count, unique, top y freq de las columnas de texto |
| `pisos["tipo"].value_counts()` | Cuántas veces sale cada valor (`normalize=True`: proporción) |
| `pisos["barrio"].nunique()` · `.unique()` | Cuántos valores distintos hay · cuáles son |
| `pisos.isna().sum()` | Cuántos datos faltan en cada columna |
| `pisos["precio"]` · `pisos[["barrio", "precio"]]` | Una columna (Series) · varias (DataFrame) |
| `pd.set_option("display.max_columns", None)` | Ver todas las columnas al usar `print()` |

**Seis preguntas al abrir un archivo:** ¿cuántas filas y columnas tiene? (`shape`) · ¿cómo son las primeras filas? (`head`) · ¿de qué tipo es cada columna? (`info`) · ¿faltan datos? (`isna().sum()`) · ¿cómo son los números? (`describe`) · ¿qué valores toman las categorías? (`value_counts`)
